```python
# 1번 통합 데이터 불러오기와 문자열 정리

from unicodedata import name

import pandas as pd # 데이터 처리

df = pd.read_csv('C:/Users/user/Downloads/integrated_student_data.csv')
# 파일이 있는 경로로 부터 데이터 불러오기

df['name'] = df['name'].str.strip() # name 열의 문자열 양쪽 공백 제거
df['email'] = df['email'].str.strip() # email 열의 앞뒤 공백 제거 후
df['email'] = df['email'].str.lower() # email 열의 소문자 변환
df['domain'] = df['email'].apply(lambda x: x.split('@')[1] if '@' in x else 'invalid')
# lambda로 domain 열 생성, 뒤에 @를 포함한 기준으로 나눈뒤 뒤 부분을 domain에 저장
df['final_score'] = df[['assignment_score', 'quiz_avg', 'project_score']].mean(axis=1)
# axis는 행, 행 방향으로 mean(평균을 구함)
df['final_score'] = df['final_score'].round(1) # final_score 값을 소수점 첫째 자리까지 반올림
df['achievement'] = df['final_score'].apply(lambda x:'우수' if x >= 85 else ('보통' if 70 <= x < 85 else '주의'))
df.head(5) # DataFrame의 상위 5행을 출력
```

1. csv 파일을 pandas로 불러온뒤
2. 공백을 제거하는 strip 명령어로 공백 제거
3. email의 앞뒤 공백을 제거하고 lower로 소문자 변경
4. lambda로 domain 생성후 @를 포함하여 나눠 0(abcd@), 1()

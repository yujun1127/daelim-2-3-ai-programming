```python
number = [1, 2, 3, 4, 5]
sum = 0
for num in number:
    sum += num
    print(sum)
print("총합:", sum)
```

```python
applicants = [94, 85, 76, 65, 50, 48]
a = 0
b = 0

print("70점 이상 합격입니다.")
for score in applicants :
    if score >= 70:
        a += 1
    else:
        b += 1
print(f'합격자는 {a}명입니다.')
print(f'불합격자는 {b}명입니다.')
```
